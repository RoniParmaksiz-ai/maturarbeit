```python
from pandas import test
import swissparlpy as spp
from swissparlpy import SwissParlClient
import pandas as pd
from datetime import datetime, time, timezone
import os

odata_client = SwissParlClient(backend="odata")
test_data_tables = odata_client.get_tables()

folder = "parlament_data"

os.makedirs(folder, exist_ok=True)


def save_transcripts_to_pkl(transcripts_data, folder, filename, date_from, date_to):
     date_from = datetime.fromisoformat(f"{year}-01-01").replace(tzinfo=timezone.utc)
     date_to = datetime.fromisoformat(f"{year + 1}-01-01").replace(tzinfo=timezone.utc)
     transcripts_data = odata_client.get_data(
        "Transcript",
        Language="DE",
        Start__gt=(date_from),
        End__lte=(date_to) 
    )
     print(f"{transcripts_data.count} rows loaded")
     df = pd.DataFrame(transcripts_data)
     pickle_path = os.path.join(folder, filename)
     df.to_pickle(pickle_path)
     print(f"Data saved to {pickle_path}")


for year in range(2000, 2001):

    filename = f"transcripts_{year}.pkl"
    pickle_path = os.path.join(folder, filename)

    if os.path.exists(pickle_path):
        print(f"Skipping {year}, already downloaded.")
        continue

    date_from = f"{year}-01-01"
    date_to = f"{year + 1}-01-01"

    print(f"\nDownloading transcripts for year {year}")


    save_transcripts_to_pkl(
         None, folder, filename, date_from, date_to
    )


files = [
    os.path.join(folder, x)
    for x in os.listdir(folder)
    if x.startswith("transcripts_") and x.endswith(".pkl")
]

if files:
    df_transcripts = pd.concat(
        [pd.read_pickle(file) for file in files],
        ignore_index=True
    )

    print(f"\nTotal transcripts: {len(df_transcripts)}")

    df_transcripts.to_pickle(
        os.path.join(folder, "all_transcripts.pkl")
    )


# finally worked now i just need to be able to open it
```
